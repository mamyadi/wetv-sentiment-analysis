## Deskripsi Proyek
Notebook ini bertujuan untuk melakukan pengambilan data (*scraping*) ulasan pengguna aplikasi streaming WeTV (`com.tencent.qqlivei18n`) secara langsung dari Google Play Store menggunakan pustaka `google-play-scraper`. 

Target perolehan data adalah minimal **10.000 ulasan** berbahasa Indonesia untuk memenuhi kriteria dataset skala besar pada evaluasi model klasifikasi sentimen multi-kelas.

## 1. Import Library
Mengimpor pustaka yang diperlukan untuk proses pengambilan data ulasan dari Google Play Store serta pengolahan dataset ke dalam format tabular.

```python
import pandas as pd
from google_play_scraper import Sort, reviews
import time

print("Pustaka berhasil diimpor.")
```

## 2. Scraping Data Ulasan WeTV
- **App ID:** `com.tencent.qqlivei18n` (ID resmi WeTV di Google Play Store)
- **Bahasa & Negara:** Indonesia (`id`)
- **Urutan:** Ulasan terbaru (`Sort.NEWEST`)
- **Target Data:** 12.000 ulasan (melebihi target minimal 10.000 data sebagai cadangan saat pembersihan ulasan kosong/duplikat).

```python
app_id = 'com.tencent.qqlivei18n'
target_count = 12000

scraped_reviews = []
continuation_token = None

print(f"Memulai proses penarikan data untuk: {app_id}")
start_time = time.time()

while len(scraped_reviews) < target_count:
    rvs, continuation_token = reviews(
        app_id,
        lang='id',
        country='id',
        sort=Sort.NEWEST,
        count=200,
        continuation_token=continuation_token
    )
    
    scraped_reviews.extend(rvs)
    
    # Cetak progres hanya tiap kelipatan 2.000 data agar output tetap bersih dan ringkas
    if len(scraped_reviews) % 2000 < 200 or len(scraped_reviews) >= target_count:
        print(f"Data terkumpul: {len(scraped_reviews)} / {target_count}")
    
    if not continuation_token:
        print("Semua ulasan yang tersedia telah diambil.")
        break
        
    time.sleep(1)

duration = round((time.time() - start_time) / 60, 2)
print(f"\nSelesai! Berhasil mengumpulkan {len(scraped_reviews)} ulasan mentah dalam waktu {duration} menit.")
```

## 3. Ekspor Data Mentah ke CSV
Menyimpan seluruh data ulasan hasil scraping beserta semua atribut ke dalam format `.csv` dengan nama `wetv-reviews-raw.csv`.

```python
# Konversi seluruh data mentah
df_raw = pd.DataFrame(scraped_reviews)

output_filename = "wetv-reviews-raw.csv"
df_raw.to_csv(output_filename, index=False, encoding='utf-8')

print(f"Penyimpanan data mentah berhasil!")
print(f"File tersimpan : {output_filename}")
print(f"Jumlah baris   : {df_raw.shape[0]}")
print(f"Jumlah kolom   : {df_raw.shape[1]}")
print(f"Daftar kolom   : {list(df_raw.columns)}")
```
